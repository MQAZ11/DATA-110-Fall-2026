```python
import matplotlib.pyplot as plt
import numpy as np
```

```python


study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

plt.plot(study_hours, exam_scores, '-ok', color='Blue')
plt.title("Relationship between hours studied and exam scores")
plt.xlabel("Hours studied")
plt.ylabel("Exam scores")
```

This graph shows that an increase in the number of hours studied leads to a higher exam score.

```python
from matplotlib.lines import lineStyles
from re import X
%matplotlib inline

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

plt.plot(days, city_a, label='city A', linestyle=('dashed'))
plt.plot(days, city_b, label='city B', linestyle=('dashdot'))
plt.title("Temperatures over one week")
plt.xlabel("day")
plt.ylabel("temperature")
plt.legend();
```

City A was generally warmer throughout the week.

```python
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]


x_indexes = np.arange(len(companies))
print(x_indexes)
plt.bar(x_indexes - 0.250, product1, width=0.25, label='Product 1')
plt.bar(x_indexes - 0.000, product2, width=0.25, label='Product 2')
plt.bar(x_indexes + 0.250, product3, width=0.25, label='Product 3')
plt.xticks(ticks=x_indexes, labels=companies)

plt.legend()
plt.show()

```
